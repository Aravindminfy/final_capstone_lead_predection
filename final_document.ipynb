### 10.3 Production Optimization Strategies

#### 10.3.1 Model Serving Optimization
```python
class OptimizedModelServer:
    def __init__(self, model_path, batch_size=32, use_gpu=False):
        self.model = self.load_model(model_path)
        self.batch_size = batch_size
        self.use_gpu = use_gpu
        self.prediction_cache = LRUCache(maxsize=1000)
    
    def batch_predict(self, data_batch):
        """Optimized batch prediction with caching"""
        batch_key = self._generate_batch_key(data_batch)
        
        if batch_key in self.prediction_cache:
            return self.prediction_cache[batch_key]
        
        # Process in optimal batch sizes
        predictions = []
        for i in range(0, len(data_batch), self.batch_size):
            batch = data_batch[i:i + self.batch_size]
            batch_pred = self.model.predict_proba(batch)
            predictions.extend(batch_pred)
        
        result = np.array(predictions)
        self.prediction_cache[batch_key] = result
        return result
    
    def async_predict(self, data):
        """Asynchronous prediction for high-throughput scenarios"""
        import asyncio
        
        async def predict_async():
            loop = asyncio.get_event_loop()
            result = await loop.run_in_executor(None, self.model.predict_proba, data)
            return result
        
        return asyncio.run(predict_async())
```

#### 10.3.2 Memory Management
```python
class MemoryOptimizedPipeline:
    def __init__(self, chunk_size=1000):
        self.chunk_size = chunk_size
        
    def process_large_dataset(self, file_path, output_path):
        """Process large datasets in chunks to manage memory"""
        chunk_results = []
        
        for chunk in pd.read_csv(file_path, chunksize=self.chunk_size):
            # Process chunk
            processed_chunk = self.preprocess_chunk(chunk)
            predictions = self.predict_chunk(processed_chunk)
            
            # Save intermediate results
            chunk_results.append(predictions)
            
            # Clear memory
            del chunk, processed_chunk
            gc.collect()
        
        # Combine results
        final_results = pd.concat(chunk_results, ignore_index=True)
        final_results.to_csv(output_path, index=False)
        
        return final_results
    
    def preprocess_chunk(self, chunk):
        """Memory-efficient preprocessing for data chunks"""
        # Apply preprocessing pipeline to chunk
        return self.preprocessing_pipeline.transform(chunk)
    
    def predict_chunk(self, chunk):
        """Memory-efficient prediction for data chunks"""
        return self.model.predict_proba(chunk)
```

## 5. Phase 2: Data Preprocessing Pipeline

### 5.1 Preprocessing Architecture

#### 5.1.1 Pipeline Design Principles
```python
class DataPreprocessor:
    def __init__(self):
        self.numerical_transformer = MinMaxScaler()
        self.categorical_transformer = OneHotEncoder(drop='first', sparse_output=False)
        self.binary_encoder = LabelEncoder()
        self.missing_value_strategy = {}
```

**Design Rationale:**
- **Modular Approach**: Separate transformers for different data types
- **Scalability**: Pipeline can handle new features without code changes
- **Reproducibility**: Consistent transformation across training and inference
- **Flexibility**: Easy to modify individual transformation steps

#### 5.1.2 Missing Value Treatment Strategy

```python
def fill_missing_values(df):
    # Numerical columns - median imputation
    numerical_cols = ['TotalVisits', 'Total Time Spent on Website', 
                      'Page Views Per Visit', 'Asymmetrique Activity Score', 
                      'Asymmetrique Profile Score']
    
    for col_name in numerical_cols:
        if col_name in df.columns:
            median_val = df[col_name].median()
            df[col_name].fillna(median_val, inplace=True)
    
    # Binary columns - default to 'No'
    binary_cols = ['Do Not Email', 'Do Not Call', 'A free copy of Mastering The Interview']
    for col_name in binary_cols:
        if col_name in df.columns:
            df[col_name].fillna('No', inplace=True)
    
    # High cardinality - mode imputation
    high_card_cols = ['Tags', 'Lead Quality', 'Lead Profile', 'Last Activity']
    for col_name in high_card_cols:
        if col_name in df.columns:
            mode_val = df[col_name].mode()[0] if len(df[col_name].mode()) > 0 else 'Unknown'
            df[col_name].fillna(mode_val, inplace=True)
```

**Missing Value Strategy Rationale:**

| Feature Type | Strategy | Rationale |
|--------------|----------|-----------|
| **Numerical** | Median Imputation | Robust to outliers, preserves distribution |
| **Binary** | Default 'No' | Business logic - assume no preference |
| **High Cardinality** | Mode Imputation | Preserves most common category |
| **Medium Cardinality** | 'Unknown' Category | Explicit handling of missing information |

### 5.2 Feature Engineering Implementation

#### 5.2.1 Binary Feature Encoding
```python
def encode_binary_columns(df, binary_cols):
    df_encoded = df.copy()
    
    for col_name in binary_cols:
        if col_name in df_encoded.columns:
            # Standardize values
            df_encoded[col_name] = df_encoded[col_name].astype(str).str.strip().str.lower()
            
            # Create binary mapping
            binary_mapping = {
                'yes': 1, 'y': 1, '1': 1, 'true': 1, 'on': 1,
                'no': 0, 'n': 0, '0': 0, 'false': 0, 'off': 0,
                'unknown': 0, 'nan': 0, 'none': 0
            }
            
            df_encoded[f"{col_name}_encoded"] = df_encoded[col_name].map(binary_mapping).fillna(0).astype(int)
    
    return df_encoded
```

**Encoding Rationale:**
- **Explicit Mapping**: Clear business logic for binary conversion
- **Default Handling**: Unknown values default to 0 (conservative approach)
- **New Column Creation**: Preserves original data for validation

#### 5.2.2 Categorical Feature Processing
```python
def create_preprocessing_pipeline(df):
    # Column categorization
    numerical_cols = ['TotalVisits', 'Total Time Spent on Website', 'Page Views Per Visit']
    categorical_cols = ['Lead Origin', 'Lead Source', 'Tags', 'Lead Quality']
    
    # Create column transformer
    preprocessor = ColumnTransformer(
        transformers=[
            ('num', MinMaxScaler(), numerical_cols),
            ('cat', OneHotEncoder(drop='first', sparse_output=False, handle_unknown='ignore'), categorical_cols)
        ],
        remainder='drop'
    )
    
    return preprocessor
```

**Transformation Strategy:**
- **MinMax Scaling**: Normalizes numerical features to [0,1] range
- **One-Hot Encoding**: Creates binary indicators for categorical values
- **Handle Unknown**: Gracefully handles new categories in production
- **Drop First**: Prevents multicollinearity in one-hot encoded features

```python
def train_models(self, cv_folds=5, scoring='roc_auc', n_jobs=-1, max_evals=20):
    cv = StratifiedKFold(n_splits=cv_folds, shuffle=True, random_state=self.random_state)
    
    for name, config in self.model_configs.items():
        model = config['model']
        param_grid = config['params']
        
        # Use RandomizedSearchCV for large parameter spaces
        search = RandomizedSearchCV(
            model, param_grid, cv=cv, scoring=scoring, n_jobs=n_jobs, 
            n_iter=max_evals, random_state=self.random_state
        )
        
        search.fit(self.X_train, self.y_train)
        self.models[name] = search.best_estimator_
        
        self.results[name] = {
            'best_params': search.best_params_,
            'best_cv_score': search.best_score_,
            'cv_std': search.cv_results_['std_test_score'][search.best_index_],
            'model': search.best_estimator_
        }
```

### 6.2 Model Evaluation Framework

#### 6.2.1 Comprehensive Evaluation Metrics
```python
def evaluate_models(self):
    evaluation_results = {}
    
    for name, model in self.models.items():
        y_pred = model.predict(self.X_test)
        y_pred_proba = model.predict_proba(self.X_test)[:, 1]
        
        metrics = {
            'Accuracy': accuracy_score(self.y_test, y_pred),
            'Precision': precision_score(self.y_test, y_pred),
            'Recall': recall_score(self.y_test, y_pred),
            'F1-Score': f1_score(self.y_test, y_pred),
            'ROC-AUC': roc_auc_score(self.y_test, y_pred_proba)
        }
        
        evaluation_results[name] = metrics
    
    # Identify best model
    results_df = pd.DataFrame(evaluation_results).T.sort_values('ROC-AUC', ascending=False)
    self.best_model_name = results_df.index[0]
    self.best_model = self.models[self.best_model_name]
```

**Evaluation Strategy Rationale:**
- **ROC-AUC**: Primary metric for ranking discrimination ability
- **Precision**: Minimize false positives in high-priority leads
- **Recall**: Maximize capture of actual conversions
- **F1-Score**: Balance between precision and recall
- **Accuracy**: Overall correctness measure

### 7.2 Data Drift Monitoring Implementation

#### 7.2.1 Statistical Drift Detection
```python
def detect_statistical_drift(reference_data, current_data, threshold=0.1):
    drift_results = {}
    
    for column in reference_data.columns:
        if reference_data[column].dtype in ['float64', 'int64']:
            # KS test for numerical features
            ks_stat, p_value = ks_2samp(reference_data[column], current_data[column])
            drift_results[column] = {
                'method': 'ks_test',
                'statistic': ks_stat,
                'p_value': p_value,
                'drift_detected': p_value < threshold
            }
        else:
            # Chi-square test for categorical features
            ref_counts = reference_data[column].value_counts()
            curr_counts = current_data[column].value_counts()
            
            # Align categories
            all_categories = set(ref_counts.index) | set(curr_counts.index)
            ref_aligned = [ref_counts.get(cat, 0) for cat in all_categories]
            curr_aligned = [curr_counts.get(cat, 0) for cat in all_categories]
            
            if sum(ref_aligned) > 0 and sum(curr_aligned) > 0:
                chi2_stat, p_value = chisquare(curr_aligned, ref_aligned)
                drift_results[column] = {
                    'method': 'chi_square',
                    'statistic': chi2_stat,
                    'p_value': p_value,
                    'drift_detected': p_value < threshold
                }
    
    return drift_results
```

#### 7.2.2 Model Performance Drift Detection
```python
def detect_performance_drift(model, reference_X, reference_y, current_X, current_y):
    # Calculate performance on reference data
    ref_predictions = model.predict_proba(reference_X)[:, 1]
    ref_auc = roc_auc_score(reference_y, ref_predictions)
    
    # Calculate performance on current data
    curr_predictions = model.predict_proba(current_X)[:, 1]
    curr_auc = roc_auc_score(current_y, curr_predictions)
    
    # Performance degradation analysis
    performance_drop = ref_auc - curr_auc
    performance_drift = performance_drop > 0.05  # 5% threshold
    
    return {
        'reference_auc': ref_auc,
        'current_auc': curr_auc,
        'performance_drop': performance_drop,
        'drift_detected': performance_drift
    }
```

## 8. Phase 5: Prediction Pipeline

### 8.1 Production Prediction Architecture

#### 8.1.1 Pipeline Class Implementation
```python
class LeadConversionPipeline:
    def __init__(self, pipeline_path: str, metadata_path: str, model_path: str):
        self.pipeline_path = pipeline_path
        self.metadata_path = metadata_path
        self.model_path = model_path
        
        self.preprocessor = None
        self.metadata = None
        self.model = None
        self.model_data = None
        
        self._load_components()
    
    def _load_components(self) -> None:
        """Load all pipeline components with error handling"""
        try:
            # Load preprocessing pipeline
            with open(self.pipeline_path, 'rb') as f:
                self.preprocessor = pickle.load(f)
            
            # Load metadata
            with open(self.metadata_path, 'rb') as f:
                self.metadata = pickle.load(f)
            
            # Load trained model with flexible key handling
            with open(self.model_path, 'rb') as f:
                self.model_data = pickle.load(f)
            
            # Extract model from various possible keys
            if isinstance(self.model_data, dict):
                for key in ['best_estimator_', 'model', 'estimator', 'best_model']:
                    if key in self.model_data:
                        self.model = self.model_data[key]
                        break
            else:
                self.model = self.model_data
            
            if not hasattr(self.model, 'predict'):
                raise ValueError("Loaded object is not a valid model")
                
        except Exception as e:
            logger.error(f"Error loading components: {e}")
            raise
```

**Component Loading Strategy:**
- **Flexible Model Loading**: Handles various model storage formats
- **Error Handling**: Comprehensive exception handling with logging
- **Validation**: Ensures loaded components are functional
- **Compatibility**: Works with different pickle formats and structures

#### 8.1.2 Data Preprocessing for Inference
```python
def _clean_data_types(self, df: pd.DataFrame) -> pd.DataFrame:
    """Clean and standardize data types for inference"""
    df_clean = df.copy()
    numerical_cols = self.metadata['numerical_features']
    categorical_cols = self.metadata['categorical_features']
    
    # Clean numerical features
    for col in numerical_cols:
        if col in df_clean.columns:
            df_clean[col] = pd.to_numeric(df_clean[col].astype(str), errors='coerce')
            # Fill missing with median from training or 0
            if df_clean[col].isna().sum() > 0:
                median_val = df_clean[col].median()
                df_clean[col].fillna(median_val if not pd.isna(median_val) else 0, inplace=True)
    
    # Clean categorical features
    for col in categorical_cols:
        if col in df_clean.columns:
            df_clean[col] = (df_clean[col].astype(str)
                           .replace(['nan', 'None', 'null'], 'Unknown')
                           .str.strip())
    
    return df_clean

def _encode_binary_features(self, df: pd.DataFrame) -> pd.DataFrame:
    """Encode binary features with business logic"""
    df_encoded = df.copy()
    binary_cols = self.metadata['binary_features']
    
    for col in binary_cols:
        if col in df_encoded.columns:
            df_encoded[col] = df_encoded[col].astype(str).str.strip().str.lower()
            
            # Comprehensive binary mapping
            binary_mapping = {
                'yes': 1, 'y': 1, '1': 1, 'true': 1, 'on': 1,
                'no': 0, 'n': 0, '0': 0, 'false': 0, 'off': 0,
                'unknown': 0, 'nan': 0, 'none': 0
            }
            
            df_encoded[f'{col}_encoded'] = (df_encoded[col]
                                          .map(binary_mapping)
                                          .fillna(0)
                                          .astype(int))
    
    return df_encoded
```

#### 8.1.3 Feature Alignment for Production
```python
def _align_features(self, X: pd.DataFrame) -> pd.DataFrame:
    """Align features to match training schema"""
    # Get expected features from model metadata
    if 'feature_names' in self.model_data:
        expected_features = self.model_data['feature_names']
    elif hasattr(self.model, 'feature_names_in_'):
        expected_features = self.model.feature_names_in_
    else:
        return X
    
    # Create aligned DataFrame with zeros for missing features
    X_aligned = pd.DataFrame(0, index=X.index, columns=expected_features)
    
    # Fill in available features
    for col in X.columns:
        if col in expected_features:
            X_aligned[col] = X[col]
    
    return X_aligned
```

**Feature Alignment Rationale:**
- **Schema Consistency**: Ensures prediction input matches training schema
- **Missing Feature Handling**: Default zero values for missing features
- **New Feature Tolerance**: Gracefully ignores unexpected features
- **Production Robustness**: Prevents prediction failures due to schema mismatches

### 8.2 Prediction Methods Implementation

#### 8.2.1 Batch Prediction Pipeline
```python
def predict_from_file(self, file_path: str, output_path: Optional[str] = None) -> Dict[str, Any]:
    """Process CSV file and generate predictions with comprehensive output"""
    # Load and preprocess data
    df = pd.read_csv(file_path)
    results = self.predict(df)
    
    # Create comprehensive results DataFrame
    results_df = results['input_data'].copy()
    results_df['Predicted_Converted'] = results['predictions']
    results_df['Lead_Score'] = results['predictions']  # Binary score
    
    # Add probability scores if available
    if results['probabilities'] is not None:
        prob_df = pd.DataFrame(
            results['probabilities'],
            columns=[f'Prob_Class_{i}' for i in range(results['probabilities'].shape[1])],
            index=results_df.index
        )
        results_df = pd.concat([results_df, prob_df], axis=1)
        
        # Business-friendly probability column
        prob_positive = results['probabilities'][:, 1]
        results_df['Conversion_Probability'] = np.round(prob_positive * 100, 2)
        
        # Lead quality categories
        results_df['Lead_Quality_Score'] = pd.cut(
            prob_positive, 
            bins=[0, 0.3, 0.6, 1.0], 
            labels=['Low', 'Medium', 'High']
        )
    
    # Add validation columns if actual labels exist
    if results['actual_labels'] is not None:
        results_df['Actual_Converted'] = results['actual_labels']
        results_df['Prediction_Correct'] = (
            results_df['Predicted_Converted'] == results_df['Actual_Converted']
        )
    
    # Save results if output path provided
    if output_path:
        results_df.to_csv(output_path, index=False)
    
    results['results_df'] = results_df
    return results
```

### 8.3 Pipeline Performance Optimization

#### 8.3.1 Caching Strategy
```python
class CachedPredictionPipeline(LeadConversionPipeline):
    def __init__(self, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.prediction_cache = {}
        self.cache_ttl = 3600  # 1 hour TTL
    
    def _get_cache_key(self, df: pd.DataFrame) -> str:
        """Generate cache key from DataFrame content"""
        return hashlib.md5(df.to_string().encode()).hexdigest()
    
    def predict_with_cache(self, df: pd.DataFrame) -> Dict[str, Any]:
        """Prediction with caching for performance"""
        cache_key = self._get_cache_key(df)
        
        # Check cache
        if cache_key in self.prediction_cache:
            cache_entry = self.prediction_cache[cache_key]
            if datetime.now() - cache_entry['timestamp'] < timedelta(seconds=self.cache_ttl):
                return cache_entry['results']
        
        # Generate new prediction
        results = self.predict(df)
        
        # Cache results
        self.prediction_cache[cache_key] = {
            'results': results,
            'timestamp': datetime.now()
        }
        
        return results
```

#### 9.1.3 Enhanced Results Processing
```python
def create_enhanced_results(results: Dict, original_df: pd.DataFrame) -> pd.DataFrame:
    """Create comprehensive results with business insights"""
    results_df = original_df.copy()
    
    # Core predictions
    results_df['Lead_Score_Prediction'] = results['predictions']
    results_df['Conversion_Prediction'] = [
        'High Potential' if pred == 1 else 'Low Potential' 
        for pred in results['predictions']
    ]
    
    # Probability analysis
    if results['probabilities'] is not None:
        prob_positive = results['probabilities'][:, 1]
        results_df['Conversion_Probability'] = np.round(prob_positive * 100, 2)
        
        # Business categorization
        results_df['Lead_Quality_Score'] = pd.cut(
            prob_positive, 
            bins=[0, 0.3, 0.6, 1.0], 
            labels=['Low', 'Medium', 'High']
        )
        
        # Priority ranking
        results_df['Priority_Rank'] = prob_positive.rank(ascending=False).astype(int)
        
        # Recommended actions
        results_df['Recommended_Action'] = results_df['Lead_Quality_Score'].map({
            'High': 'Immediate Follow-up',
            'Medium': 'Nurture Campaign',
            'Low': 'Monitor'
        })
    
    return results_df

def generate_business_summary(results_df: pd.DataFrame) -> Dict:
    """Generate comprehensive business summary"""
    total_records = len(results_df)
    high_potential = sum(results_df['Lead_Score_Prediction'])
    
    summary = {
        'total_records': total_records,
        'high_potential_leads': int(high_potential),
        'low_potential_leads': total_records - int(high_potential),
        'conversion_rate_prediction': round((high_potential / total_records) * 100, 2),
        'top_10_leads': results_df.nlargest(10, 'Conversion_Probability')[['Prospect ID', 'Conversion_Probability']].to_dict('records') if 'Conversion_Probability' in results_df.columns else [],
        'priority_distribution': results_df['Lead_Quality_Score'].value_counts().to_dict() if 'Lead_Quality_Score' in results_df.columns else {}
    }
    
    return summary
```

## 10. Technical Implementation Details

### 10.1 Data Pipeline Architecture

#### 10.1.1 ETL Process Implementation
```python
class DataPipeline:
    def __init__(self, config):
        self.config = config
        self.logger = logging.getLogger(__name__)
    
    def extract(self, source_path: str) -> pd.DataFrame:
        """Extract data from various sources"""
        if source_path.endswith('.csv'):
            df = pd.read_csv(source_path)
        elif source_path.endswith('.xlsx'):
            df = pd.read_excel(source_path)
        else:
            raise ValueError(f"Unsupported file format: {source_path}")
        
        self.logger.info(f"Extracted {len(df)} records from {source_path}")
        return df
    
    def transform(self, df: pd.DataFrame) -> pd.DataFrame:
        """Apply comprehensive data transformations"""
        # Data validation
        df_validated = self.validate_data_quality(df)
        
        # Missing value treatment
        df_imputed = self.handle_missing_values(df_validated)
        
        # Feature engineering
        df_engineered = self.engineer_features(df_imputed)
        
        # Data type conversion
        df_typed = self.convert_data_types(df_engineered)
        
        return df_typed
    
    def load(self, df: pd.DataFrame, output_path: str) -> None:
        """Load processed data to destination"""
        df.to_csv(output_path, index=False)
        self.logger.info(f"Loaded {len(df)} records to {output_path}")
```

#### 10.1.2 Feature Engineering Framework
```python
class FeatureEngineer:
    def __init__(self):
        self.engineered_features = []
    
    def create_engagement_features(self, df: pd.DataFrame) -> pd.DataFrame:
        """Create engagement-based features"""
        df_eng = df.copy()
        
        # Composite engagement score
        if all(col in df.columns for col in ['TotalVisits', 'Total Time Spent on Website', 'Page Views Per Visit']):
            df_eng['Engagement_Score'] = (
                df_eng['Total Time Spent on Website'].fillna(0) * 0.6 +
                df_eng['TotalVisits'].fillna(0) * 100 * 0.3 +
                df_eng['Page Views Per Visit'].fillna(0) * 50 * 0.1
            )
            self.engineered_features.append('Engagement_Score')
        
        # Visit intensity
        if 'TotalVisits' in df.columns:
            df_eng['Visit_Intensity'] = pd.cut(
                df_eng['TotalVisits'].fillna(0),
                bins=[0, 1, 5, 10, float('inf')],
                labels=['None', 'Low', 'Medium', 'High']
            )
            self.engineered_features.append('Visit_Intensity')
        
        return df_eng
    
    def create_temporal_features(self, df: pd.DataFrame) -> pd.DataFrame:
        """Create time-based features if timestamps available"""
        df_temporal = df.copy()
        
        # This would be implemented if date columns were available
        # Example: days_since_first_contact, days_since_last_activity
        
        return df_temporal
    
    def create_interaction_features(self, df: pd.DataFrame) -> pd.DataFrame:
        """Create interaction features between important variables"""
        df_int = df.copy()
        
        # Lead Source × Quality interaction
        if all(col in df.columns for col in ['Lead Source', 'Lead Quality']):
            df_int['Source_Quality_Interaction'] = (
                df_int['Lead Source'].astype(str) + '_' + 
                df_int['Lead Quality'].astype(str)
            )
            self.engineered_features.append('Source_Quality_Interaction')
        
        return df_int
```

### 10.2 Model Training Framework

#### 10.2.1 Cross-Validation Strategy
```python
class ModelValidator:
    def __init__(self, cv_folds=5, random_state=42):
        self.cv_folds = cv_folds
        self.random_state = random_state
    
    def stratified_cv_split(self, X, y):
        """Create stratified cross-validation splits"""
        skf = StratifiedKFold(
            n_splits=self.cv_folds, 
            shuffle=True, 
            random_state=self.random_state
        )
        return skf.split(X, y)
    
    def time_series_cv_split(self, X, y, time_column):
        """Create time-aware cross-validation splits"""
        # Sort by time
        sorted_indices = X[time_column].argsort()
        X_sorted = X.iloc[sorted_indices]
        y_sorted = y.iloc[sorted_indices]
        
        # Create time-based splits
        n_samples = len(X_sorted)
        test_size = n_samples // self.cv_folds
        
        for i in range(self.cv_folds):
            test_start = i * test_size
            test_end = (i + 1) * test_size if i < self.cv_folds - 1 else n_samples
            
            train_indices = list(range(0, test_start))
            test_indices = list(range(test_start, test_end))
            
            yield train_indices, test_indices
    
    def validate_model(self, model, X, y, cv_method='stratified'):
        """Comprehensive model validation"""
        if cv_method == 'stratified':
            cv_splits = self.stratified_cv_split(X, y)
        else:
            cv_splits = self.time_series_cv_split(X, y, 'timestamp')
        
        cv_scores = []
        feature_importances = []
        
        for train_idx, val_idx in cv_splits:
            X_train, X_val = X.iloc[train_idx], X.iloc[val_idx]
            y_train, y_val = y.iloc[train_idx], y.iloc[val_idx]
            
            # Train model
            model.fit(X_train, y_train)
            
            # Predict and score
            y_pred_proba = model.predict_proba(X_val)[:, 1]
            score = roc_auc_score(y_val, y_pred_proba)
            cv_scores.append(score)
            
            # Collect feature importance
            if hasattr(model, 'feature_importances_'):
                feature_importances.append(model.feature_importances_)
        
        return {
            'cv_scores': cv_scores,
            'mean_score': np.mean(cv_scores),
            'std_score': np.std(cv_scores),
            'feature_importance_mean': np.mean(feature_importances, axis=0) if feature_importances else None
        }
```

#### 10.2.2 Hyperparameter Optimization
```python
class HyperparameterOptimizer:
    def __init__(self, model_class, param_distributions, n_iter=50):
        self.model_class = model_class
        self.param_distributions = param_distributions
        self.n_iter = n_iter
    
    def bayesian_optimization(self, X_train, y_train, X_val, y_val):
        """Bayesian optimization for hyperparameter tuning"""
        from skopt import gp_minimize
        from skopt.space import Real, Integer, Categorical
        
        def objective(params):
            # Create model with parameters
            param_dict = dict(zip(self.param_distributions.keys(), params))
            model = self.model_class(**param_dict)
            
            # Train and evaluate
            model.fit(X_train, y_train)
            y_pred_proba = model.predict_proba(X_val)[:, 1]
            score = roc_auc_score(y_val, y_pred_proba)
            
            # Return negative score for minimization
            return -score
        
        # Define search space
        dimensions = []
        for param_name, param_values in self.param_distributions.items():
            if isinstance(param_values[0], int):
                dimensions.append(Integer(min(param_values), max(param_values)))
            elif isinstance(param_values[0], float):
                dimensions.append(Real(min(param_values), max(param_values)))
            else:
                dimensions.append(Categorical(param_values))
        
        # Run optimization
        result = gp_minimize(
            func=objective,
            dimensions=dimensions,
            n_calls=self.n_iter,
            random_state=42
        )
        
        # Return best parameters
        best_params = dict(zip(self.param_distributions.keys(), result.x))
        return best_params, -result.fun
    
    def grid_search_optimization(self, X_train, y_train, cv_folds=5):
        """Traditional grid search with cross-validation"""
        from sklearn.model_selection import GridSearchCV
        
        model = self.model_class()
        grid_search = GridSearchCV(
            model,
            self.param_distributions,
            cv=cv_folds,
            scoring='roc_auc',
            n_jobs=-1,
            verbose=1
        )
        
        grid_search.fit(X_train, y_train)
        
        return grid_search.best_params_, grid_search.best_score_
